```python
!pip install gradio pandas -q

import sqlite3
import pandas as pd
import gradio as gr
from datetime import datetime

DB_NAME = "/content/database.db"

# 1. Inicialização do Banco de Dados
def init_db():
    with sqlite3.connect(DB_NAME) as conn:
        cursor = conn.cursor()

        cursor.execute("""
            CREATE TABLE IF NOT EXISTS salas (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                bloco TEXT NOT NULL,
                numero TEXT UNIQUE NOT NULL,
                capacidade INTEGER NOT NULL,
                tipo TEXT NOT NULL,
                recursos TEXT
            )
        """)

        cursor.execute("""
            CREATE TABLE IF NOT EXISTS usuarios (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                nome TEXT NOT NULL,
                email TEXT UNIQUE NOT NULL,
                senha TEXT NOT NULL,
                papel TEXT NOT NULL
            )
        """)

        cursor.execute("""
            CREATE TABLE IF NOT EXISTS reservas (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                sala_numero TEXT NOT NULL,
                professor_nome TEXT NOT NULL,
                data TEXT NOT NULL,
                periodo TEXT NOT NULL,
                disciplina TEXT NOT NULL
            )
        """)

        # Popula Salas (1026 a 1040)
        cursor.execute("SELECT COUNT(*) FROM salas WHERE tipo = 'Sala de Aula'")
        if cursor.fetchone()[0] == 0:
            salas_aula = [('Bloco Central', f'Sala {i}', 40, 'Sala de Aula', 'Projetor, Ar-Condicionado, Lousa Digital') for i in range(1026, 1041)]
            cursor.executemany("INSERT INTO salas (bloco, numero, capacidade, tipo, recursos) VALUES (?, ?, ?, ?, ?)", salas_aula)

        # Popula Labs (1 a 8)
        cursor.execute("SELECT COUNT(*) FROM salas WHERE tipo = 'Laboratório'")
        if cursor.fetchone()[0] == 0:
            laboratorios = [('Bloco de Tecnologia', f'Lab {i}', 30, 'Laboratório', '30 PCs de Alta Performance, Projetor, Ar-Condicionado') for i in range(1, 9)]
            cursor.executemany("INSERT INTO salas (bloco, numero, capacidade, tipo, recursos) VALUES (?, ?, ?, ?, ?)", laboratorios)

        # Usuários Padrão
        cursor.execute("SELECT COUNT(*) FROM usuarios")
        if cursor.fetchone()[0] == 0:
            cursor.execute("INSERT INTO usuarios (nome, email, senha, papel) VALUES (?, ?, ?, ?)", ("Prof. Carlos", "prof@unifil.br", "123456", "Professor"))
            cursor.execute("INSERT INTO usuarios (nome, email, senha, papel) VALUES (?, ?, ?, ?)", ("Coord. Ana", "coord@unifil.br", "123456", "Coordenador"))

        conn.commit()

init_db()

# 2. Funções de Consulta e Regras de Negócio
def obter_metricas():
    with sqlite3.connect(DB_NAME) as conn:
        c = conn.cursor()
        c.execute("SELECT COUNT(*) FROM salas WHERE tipo = 'Sala de Aula'")
        total_salas = c.fetchone()[0]
        c.execute("SELECT COUNT(*) FROM salas WHERE tipo = 'Laboratório'")
        total_labs = c.fetchone()[0]
        c.execute("SELECT COUNT(*) FROM reservas")
        total_reservas = c.fetchone()[0]

    html = f"""
    <div style="display: flex; gap: 15px; margin-bottom: 15px;">
        <div style="flex: 1; background: #ffffff; border-top: 5px solid #F15A24; padding: 18px; border-radius: 10px; box-shadow: 0 4px 12px rgba(241, 90, 36, 0.15);">
            <span style="font-size: 11px; color: #F15A24; font-weight: 800; text-transform: uppercase; letter-spacing: 0.5px;">Salas de Aula</span>
            <h2 style="margin: 5px 0 0 0; color: #D84A1B; font-size: 32px; font-weight: bold;">{total_salas}</h2>
            <small style="color: #666;">Bloco Central (1026 - 1040)</small>
        </div>
        <div style="flex: 1; background: #ffffff; border-top: 5px solid #F15A24; padding: 18px; border-radius: 10px; box-shadow: 0 4px 12px rgba(241, 90, 36, 0.15);">
            <span style="font-size: 11px; color: #F15A24; font-weight: 800; text-transform: uppercase; letter-spacing: 0.5px;">Laboratórios</span>
            <h2 style="margin: 5px 0 0 0; color: #F15A24; font-size: 32px; font-weight: bold;">{total_labs}</h2>
            <small style="color: #666;">Bloco de Tecnologia (1 - 8)</small>
        </div>
        <div style="flex: 1; background: #ffffff; border-top: 5px solid #FF8C00; padding: 18px; border-radius: 10px; box-shadow: 0 4px 12px rgba(255, 140, 0, 0.15);">
            <span style="font-size: 11px; color: #FF8C00; font-weight: 800; text-transform: uppercase; letter-spacing: 0.5px;">Agendamentos Ativos</span>
            <h2 style="margin: 5px 0 0 0; color: #D84A1B; font-size: 32px; font-weight: bold;">{total_reservas}</h2>
            <small style="color: #666;">Reservas confirmadas</small>
        </div>
    </div>
    """
    return html

def fazer_login(email, senha):
    if not email or not senha:
        return "⚠️ Preencha e-mail e senha.", gr.update(visible=True), gr.update(visible=False), "", "", gr.update(visible=False), ""

    with sqlite3.connect(DB_NAME) as conn:
        cursor = conn.cursor()
        cursor.execute("SELECT nome, papel FROM usuarios WHERE email = ? AND senha = ?", (email.strip().lower(), senha))
        user = cursor.fetchone()

        if user:
            nome, papel = user
            is_coord = (papel == "Coordenador")
            msg_top = f"👤 Conectado como: **{nome}** ({papel})"
            return "", gr.update(visible=False), gr.update(visible=True), msg_top, papel, gr.update(visible=is_coord), nome
        else:
            return "❌ Credenciais inválidas. Verifique e-mail e senha.", gr.update(visible=True), gr.update(visible=False), "", "", gr.update(visible=False), ""

def cadastrar_usuario(nome, email, senha, papel):
    if not nome or not email or not senha:
        return "⚠️ Preencha todos os campos do formulário."
    with sqlite3.connect(DB_NAME) as conn:
        cursor = conn.cursor()
        try:
            cursor.execute("INSERT INTO usuarios (nome, email, senha, papel) VALUES (?, ?, ?, ?)", (nome, email.strip().lower(), senha, papel))
            conn.commit()
            return f"✅ Conta criada com sucesso para {nome}! Acesse a aba 'Entrar no Sistema'."
        except sqlite3.IntegrityError:
            return "⚠️ Este e-mail já está em uso."

def buscar_salas_filtradas(filtro_tipo, busca_nome):
    query = "SELECT id AS ID, bloco AS Bloco, numero AS 'Número/Nome', capacidade AS 'Capacidade (Alunos)', tipo AS Tipo, recursos AS Recursos FROM salas WHERE 1=1"
    params = []

    if filtro_tipo != "Todos":
        query += " AND tipo = ?"
        params.append(filtro_tipo)
    if busca_nome:
        query += " AND numero LIKE ?"
        params.append(f"%{busca_nome}%")

    query += " ORDER BY tipo, id"
    with sqlite3.connect(DB_NAME) as conn:
        return pd.read_sql_query(query, conn, params=params)

def listar_opcoes_salas():
    with sqlite3.connect(DB_NAME) as conn:
        c = conn.cursor()
        c.execute("SELECT numero FROM salas ORDER BY tipo, numero")
        return [r[0] for r in c.fetchall()]

def realizar_reserva(nome_prof, sala, data, periodo, disciplina):
    if not sala or not data or not disciplina:
        return "⚠️ Preencha todos os campos da reserva.", listar_reservas()

    with sqlite3.connect(DB_NAME) as conn:
        c = conn.cursor()
        c.execute("SELECT id FROM reservas WHERE sala_numero = ? AND data = ? AND periodo = ?", (sala, data, periodo))
        if c.fetchone():
            return f"❌ CHOQUE DE HORÁRIO: A {sala} já está reservada no dia {data} no período da {periodo}.", listar_reservas()

        c.execute("INSERT INTO reservas (sala_numero, professor_nome, data, periodo, disciplina) VALUES (?, ?, ?, ?, ?)",
                  (sala, nome_prof, data, periodo, disciplina))
        conn.commit()
        return f"✅ Reserva confirmada para a {sala} em {data} ({periodo})!", listar_reservas()

def listar_reservas():
    with sqlite3.connect(DB_NAME) as conn:
        return pd.read_sql_query("SELECT id AS ID, sala_numero AS 'Sala/Lab', professor_nome AS Professor, data AS Data, periodo AS Período, disciplina AS Disciplina FROM reservas ORDER BY data DESC", conn)

def cadastrar_sala(papel, bloco, numero, capacidade, tipo, recursos):
    if papel != "Coordenador":
        return "❌ Acesso negado.", buscar_salas_filtradas("Todos", ""), gr.update(choices=listar_opcoes_salas())
    with sqlite3.connect(DB_NAME) as conn:
        cursor = conn.cursor()
        try:
            cursor.execute("INSERT INTO salas (bloco, numero, capacidade, tipo, recursos) VALUES (?, ?, ?, ?, ?)",
                           (bloco, numero, capacidade, tipo, recursos))
            conn.commit()
            return f"✅ {tipo} '{numero}' cadastrada com sucesso!", buscar_salas_filtradas("Todos", ""), gr.update(choices=listar_opcoes_salas())
        except sqlite3.IntegrityError:
            return f"⚠️ A sala ou lab '{numero}' já existe no banco.", buscar_salas_filtradas("Todos", ""), gr.update(choices=listar_opcoes_salas())

def excluir_sala(papel, sala_id):
    if papel != "Coordenador":
        return "❌ Acesso negado.", buscar_salas_filtradas("Todos", "")
    with sqlite3.connect(DB_NAME) as conn:
        cursor = conn.cursor()
        cursor.execute("DELETE FROM salas WHERE id = ?", (int(sala_id),))
        conn.commit()
        return f"🗑️ Registro ID {sala_id} removido.", buscar_salas_filtradas("Todos", "")

# 3. CSS 100% Laranja (Sem Azul)
css_unifil_exclusivo_laranja = """
:root {
    --primary-500: #F15A24 !important;
    --primary-600: #D84A1B !important;
    --body-text-color: #222222 !important;
}
body { background-color: #fcf8f6; font-family: 'Segoe UI', Roboto, sans-serif; }

/* Header com Degradê de Laranjas */
.header-bar {
    background: linear-gradient(135deg, #F15A24 0%, #D84A1B 50%, #B3360E 100%);
    color: white;
    padding: 22px 28px;
    border-radius: 12px;
    box-shadow: 0 6px 18px rgba(241, 90, 36, 0.3);
    border-bottom: 5px solid #8E2706;
    display: flex;
    justify-content: space-between;
    align-items: center;
    margin-bottom: 20px;
}

/* Botões em Laranja Destaque */
.btn-orange {
    background: #F15A24 !important;
    color: white !important;
    font-weight: bold !important;
    border: none !important;
    transition: all 0.2s ease !important;
    box-shadow: 0 4px 10px rgba(241, 90, 36, 0.3) !important;
}
.btn-orange:hover {
    background: #D84A1B !important;
    transform: translateY(-1px);
    box-shadow: 0 6px 14px rgba(241, 90, 36, 0.45) !important;
}

/* Abas Selecionadas em Laranja */
.tabs button.selected {
    border-bottom: 3px solid #F15A24 !important;
    color: #F15A24 !important;
    font-weight: bold !important;
}
"""

# 4. Interface do Aplicativo
with gr.Blocks(css=css_unifil_exclusivo_laranja, title="UniFil - Gestão de Salas") as app:

    # Header em Laranja Exclusivo
    gr.HTML("""
        <div class="header-bar">
            <div>
                <h1 style="margin:0; font-size: 26px; font-weight: 800; letter-spacing: -0.5px; text-shadow: 0 2px 4px rgba(0,0,0,0.2);">🎓 Portal UniFil | Gestão de Salas</h1>
                <p style="margin: 4px 0 0 0; opacity: 0.95; font-size: 14px; font-weight: 500;">Sistema Oficial de Controle de Espaços Físicos e Laboratórios</p>
            </div>
            <div style="text-align: right;">
                <span style="background: rgba(255,255,255,0.2); color: #ffffff; padding: 6px 14px; border-radius: 20px; font-size: 12px; font-weight: bold; letter-spacing: 0.5px; box-shadow: 0 2px 6px rgba(0,0,0,0.15);">UniFil Londrina</span>
            </div>
        </div>
    """)

    # Estados Globais
    papel_usuario_state = gr.State("")
    nome_usuario_state = gr.State("")

    # ================= CONTEXTO 1: AUTENTICAÇÃO =================
    with gr.Column(visible=True) as secao_auth:
        with gr.Row():
            with gr.Column(scale=1):
                gr.Markdown("## 🔐 Autenticação Institucional")
                gr.Markdown("Entre com a sua conta da UniFil para navegar pelas salas e agendar horários.")

                with gr.Tabs():
                    with gr.Tab("Entrar no Sistema"):
                        in_email = gr.Textbox(label="E-mail Institucional", value="prof@unifil.br")
                        in_senha = gr.Textbox(label="Senha", type="password", value="123456")
                        btn_login = gr.Button("Entrar no Sistema", elem_classes=["btn-orange"])
                        msg_auth_err = gr.Textbox(label="Status", interactive=False)

                        gr.Markdown("---")
                        gr.Markdown("💡 **Credenciais de Teste:**")
                        gr.Markdown("- **Professor:** `prof@unifil.br` | Senha: `123456`")
                        gr.Markdown("- **Coordenador:** `coord@unifil.br` | Senha: `123456`")

                    with gr.Tab("Criar Nova Conta"):
                        in_nome_cad = gr.Textbox(label="Nome Completo")
                        in_email_cad = gr.Textbox(label="E-mail Institucional")
                        in_senha_cad = gr.Textbox(label="Senha", type="password")
                        in_papel_cad = gr.Radio(["Professor", "Coordenador"], value="Professor", label="Função")
                        btn_cad_user = gr.Button("Cadastrar Conta", elem_classes=["btn-orange"])
                        msg_cad_status = gr.Textbox(label="Status do Cadastro", interactive=False)

            with gr.Column(scale=1):
                gr.Markdown("### 🟠 Espaços Acadêmicos UniFil")
                gr.Markdown("""
                Plataforma unificada para gerenciamento das infraestruturas dos blocos:

                - 🏫 **Salas 1026 a 1040**: Salas do Bloco Central climatizadas com datashow e capacidade para 40 alunos.
                - 💻 **Laboratórios 1 a 8**: Laboratórios do Bloco de Tecnologia com 30 computadores e projetor.
                - 📅 **Agendamento em Tempo Real**: Evite conflitos de horário com validação automática de reservas.
                """)

    # ================= CONTEXTO 2: APLICAÇÃO PRINCIPAL (PÓS-LOGIN) =================
    with gr.Column(visible=False) as secao_sistema:

        # Barra de Status do Usuário
        with gr.Row():
            user_info_txt = gr.Markdown("👤 Conectado")
            btn_logout = gr.Button("🚪 Sair", size="sm")

        # Cartões de Métricas (KPIs Laranja)
        dashboard_kpis = gr.HTML(obter_metricas())

        # NAVEGAÇÃO PRINCIPAL EM ABAS
        with gr.Tabs() as abas_sistema:

            # --- ABA 1: CONSULTA DE SALAS ---
            with gr.Tab("🏢 Consultar Salas e Labs"):
                gr.Markdown("### Pesquisa de Espaços Físicos")
                with gr.Row():
                    f_tipo = gr.Dropdown(["Todos", "Sala de Aula", "Laboratório"], value="Todos", label="Filtrar por Tipo")
                    f_nome = gr.Textbox(label="Buscar por Número/Nome", placeholder="Ex: 1028 ou Lab 3")
                    btn_filtrar = gr.Button("🔍 Filtrar", elem_classes=["btn-orange"], size="sm")

                grid_salas = gr.Dataframe(value=buscar_salas_filtradas("Todos", ""), interactive=False)

            # --- ABA 2: RESERVA DE HORÁRIOS ---
            with gr.Tab("📅 Agendamento de Horários"):
                gr.Markdown("### Efetuar Nova Reserva de Sala")
                with gr.Row():
                    with gr.Column():
                        drop_salas_reserva = gr.Dropdown(choices=listar_opcoes_salas(), label="Selecione a Sala ou Laboratório")
                        in_data_reserva = gr.Textbox(label="Data da Aula/Evento", value=datetime.today().strftime('%Y-%m-%d'), placeholder="YYYY-MM-DD")
                        in_periodo_reserva = gr.Radio(["Manhã", "Tarde", "Noite"], value="Noite", label="Período")
                        in_disc_reserva = gr.Textbox(label="Disciplina / Motivo", placeholder="Ex: Algoritmos e Estrutura de Dados")
                        btn_confirmar_reserva = gr.Button("Confirmar Agendamento", elem_classes=["btn-orange"])
                        msg_reserva_status = gr.Textbox(label="Status da Reserva", interactive=False)

                    with gr.Column():
                        gr.Markdown("### 📋 Horários Já Agendados")
                        grid_reservas = gr.Dataframe(value=listar_reservas(), interactive=False)

            # --- ABA 3: PAINEL DO COORDENADOR (EXCLUSIVO) ---
            with gr.Tab("⚙️ Painel do Coordenador", visible=False) as aba_coordenador:
                gr.Markdown("### Gestão de Infraestrutura (Restrito a Coordenadores)")

                with gr.Row():
                    with gr.Column():
                        gr.Markdown("#### ➕ Cadastrar Novo Espaço")
                        c_bloco = gr.Textbox(label="Bloco", value="Bloco Central")
                        c_num = gr.Textbox(label="Número/Nome (ex: Sala 1041 ou Lab 9)")
                        c_cap = gr.Number(label="Capacidade", value=40)
                        c_tipo = gr.Dropdown(["Sala de Aula", "Laboratório"], value="Sala de Aula", label="Tipo")
                        c_rec = gr.Textbox(label="Recursos Disponíveis", value="Projetor, Ar-Condicionado")
                        btn_cad_sala = gr.Button("Cadastrar Sala/Lab", elem_classes=["btn-orange"])

                    with gr.Column():
                        gr.Markdown("#### 🗑️ Remover Espaço")
                        c_id_del = gr.Number(label="ID do Espaço (ver primeira coluna da tabela)")
                        btn_del_sala = gr.Button("Excluir Definitivamente", variant="stop")

                msg_admin_status = gr.Textbox(label="Status Administrativo", interactive=False)

    # ================= VINCULAÇÃO DE EVENTOS =================
    btn_cad_user.click(fn=cadastrar_usuario, inputs=[in_nome_cad, in_email_cad, in_senha_cad, in_papel_cad], outputs=msg_cad_status)

    btn_login.click(
        fn=fazer_login,
        inputs=[in_email, in_senha],
        outputs=[msg_auth_err, secao_auth, secao_sistema, user_info_txt, papel_usuario_state, aba_coordenador, nome_usuario_state]
    )

    btn_logout.click(fn=lambda: (gr.update(visible=True), gr.update(visible=False)), outputs=[secao_auth, secao_sistema])

    # Eventos de Filtro e Busca
    btn_filtrar.click(fn=buscar_salas_filtradas, inputs=[f_tipo, f_nome], outputs=grid_salas)
    f_tipo.change(fn=buscar_salas_filtradas, inputs=[f_tipo, f_nome], outputs=grid_salas)

    # Evento de Reserva
    btn_confirmar_reserva.click(
        fn=realizar_reserva,
        inputs=[nome_usuario_state, drop_salas_reserva, in_data_reserva, in_periodo_reserva, in_disc_reserva],
        outputs=[msg_reserva_status, grid_reservas]
    )

    # Eventos Admin
    btn_cad_sala.click(
        fn=cadastrar_sala,
        inputs=[papel_usuario_state, c_bloco, c_num, c_cap, c_tipo, c_rec],
        outputs=[msg_admin_status, grid_salas, drop_salas_reserva]
    )
    btn_del_sala.click(
        fn=excluir_sala,
        inputs=[papel_usuario_state, c_id_del],
        outputs=[msg_admin_status, grid_salas]
    )

app.launch(share=True, inline=False)
```
